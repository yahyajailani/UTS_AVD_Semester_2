### **Outliers Values**

Pada tahap ini dilakukan pengecekan nilai ekstrem (outliers) pada kolom numerik untuk menjaga akurasi analisis. Apabila data memiliki nilai yang terlalu tinggi atau terlalu rendah terhadap nilai toleransi nya maka data tersebut bisa dikatakan memiliki outliers dan harus ditangani.

```python
results = []

cols = df.select_dtypes(include=['float64', 'int64'])

for col in cols:
  q1 = df[col].quantile(0.25)
  q3 = df[col].quantile(0.75)
  iqr = q3 - q1
  lower_bound = q1 - 1.5*iqr
  upper_bound = q3 + 1.5*iqr
  outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]
  percent_outliers = (len(outliers)/len(df))*100
  results.append({'Kolom': col, 'Persentase Outliers': percent_outliers})

#Dataframe dari list hasil

results_df = pd.DataFrame(results)
results_df.set_index('Kolom', inplace=True)
results_df = results_df.rename_axis(None, axis=0).rename_axis('Kolom', axis=1)

#Tampilkan dataframe

display(results_df)
```

Kode ini membantu mengidentifikasi seberapa besar ketidakwajaran data (outliers) pada tiap variabel, sehingga dapat menjadi dasar untuk menentukan apakah data perlu dibersihkan atau ditangani sebelum analisis lebih lanjut dilakukan

Terdapat Outliers pada kolom **depression_label** sebesar **2.583333**, maka kolom tersebut perlu ditangani agar bisa masuk ketahap analisis lanjutan.

```python
columns_to_impute = ["depression_label"]

for col in columns_to_impute:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    # Menggunakan .loc[] agar tidak muncul SettingWithCopyWarning
    df.loc[:, col] = df[col].clip(lower=lower_bound, upper=upper_bound)
```

Identifikasi pencilan menunjukkan bahwa variabel **depression_label** memiliki tingkat anomali paling tinggi (2.5%), sedangkan **age, daily_social_media_hours, sleep_hours, screen_time_before_sleep, academic_performance, physical_activity, stress_level, anxiety_level, addiction_level** menunjukkan konsistensi data yang sempurna (0.00% outliers). Tindakan lebih lanjut difokuskan pada validasi data ekstrem pada jumlah bangunan rusak untuk memastikan apakah nilai tersebut merupakan anomali teknis atau representasi bencana skala besar

```python
results = []

cols = df.select_dtypes(include=['float64', 'int64'])

for col in cols:
  q1 = df[col].quantile(0.25)
  q3 = df[col].quantile(0.75)
  iqr = q3 - q1
  lower_bound = q1 - 1.5*iqr
  upper_bound = q3 + 1.5*iqr
  outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]
  percent_outliers = (len(outliers)/len(df))*100
  results.append({'Kolom': col, 'Persentase Outliers': percent_outliers})

# Dataframe dari list hasil
results_df = pd.DataFrame(results)
results_df.set_index('Kolom', inplace=True)
results_df = results_df.rename_axis(None, axis=0).rename_axis('Kolom', axis=1)

# Tampilkan dataframe
display(results_df)
```

Identifikasi setelah pembersihan menunjukkan bahwa seluruh variabel telah mencapai tingkat persentase outliers 0.0%. Hal ini mengkonfirmasi bahwa distribusi data kini sudah berada dalam rentang normal, sehingga integritas data terjaga dan siap untuk digunakan dalam tahap analisis statistik maupun pemodelan lebih lanjut tanpa adanya bias dari nilai ekstrem

### **Relationship(Hubungan)**

Aktivitas: Menganalisis nilai korelasi antar seluruh variabel numerik (seperti daily_social_media_hours, sleep_hours, stress_level, anxiety_level, dan addiction_level) secara simultan.

Tujuan: Mengidentifikasi kekuatan dan arah hubungan (positif atau negatif) antar berbagai indikator kebiasaan digital dan kesehatan mental untuk menemukan pola gangguan psikologis yang paling dominan pada remaja.

Visualisasi: Heatmap

Alasan: Heatmap adalah pilihan terbaik untuk menampilkan matriks korelasi yang kompleks karena menggunakan gradasi warna (seperti merah untuk korelasi positif kuat dan biru untuk korelasi negatif kuat) sehingga angka korelasi yang padat bisa dibaca dan dipahami secara instan.

```python
cols = ['daily_social_media_hours', 'sleep_hours', 'screen_time_before_sleep', 'stress_level']
korelasi = df[cols].corr()

plt.figure(figsize=(10, 8))
# Membuat heatmap korelasi sesuai dengan struktur contohmu
sns.heatmap(
    korelasi,
    annot=True,
    cmap='viridis',
    fmt='.2f',
    linewidths=0.5
)

plt.title('Matriks Korelasi Faktor Perilaku Digital dan Stres Remaja', fontsize=14)
plt.show()
```

**Insight**
- **Krisis Tidur:** Terdapat korelasi negatif kuat antara durasi media sosial (**daily_social_media_hours**) dan paparan layar sebelum tidur (**screen_time_before_sleep**) dengan jam tidur (**sleep_hours**). Penggunaan gawai intensif terbukti memotong porsi istirahat.
- **Efek Akademik:** Ketergantungan gawai (**addiction_level**) berbanding terbalik dengan prestasi belajar (**academic_performance**).
- **Faktor Pelindung:** Aktivitas fisik terbukti memiliki korelasi negatif dengan tingkat stres dan kecemasan, mengonfirmasi perannya sebagai peredam risiko kesehatan mental.
